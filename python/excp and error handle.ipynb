```python
try:
    num = int(input("Enter the number: "))
    print("You entered: ", num)
    res = 35/num
    print("The result is: ", res)
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
except ValueError:
    print("Error: Please enter a valid integer.")
```
