```python
# regex (regular expression)
import re
text = "Hello Python"
x= re.search("Python", text)    #string pattern return kahi bhi mile
print(x)
if x:
    print("found")

x= re.match("Python", text)    #string ke staart me check karega
print(x)

text = "cat bat rat mat"
result = re.findall("at", text)
print(result)

text = "I like Java"
new_text = re.sub("Java", "Python", text)
print(new_text)

email = "test@gmail.com"
pattern = r'^\w+@\w+\.\w+$'
if re.match(pattern, email):
    print("Valid Email")
else:
    print("Invalid Email")

    

text = "My number is 9876543219"
match = re.search(r'(\d{10})', text)
if match:
    print(match.group())
```

```python
import pandas as pd

df = pd.DataFrame({
    'date': ['2024-01-01', '2024-01-02', '2024-01-03', '2024-01-04', '2024-01-05'],
    'sales': [100, 200, 150, 300, 250]
})

df['date'] = pd.to_datetime(df['date'])

df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
df['day'] = df['date'].dt.day
df['day_name'] = df['date'].dt.day_name()

#Filtering dates
filtered = df[df['date'] > '2024-01-02']

#Create date range
date_range = pd.date_range(start='2024-01-01', periods=5, freq='D')

#Set index for time series
df.set_index('date', inplace=True)

#Resampling (monthly sum)
resampled = df.resample('ME').sum(numeric_only=True)

#Time difference (Timedelta)
df['days_diff'] = df.index - df.index[0]

# Shift (previous value)
df['prev_sales'] = df['sales'].shift(1)

# Rolling average
df['rolling_avg'] = df['sales'].rolling(3).mean()

#Date formatting
df['formatted_date'] = df.index.strftime('%d-%m-%Y')



print("Original DataFrame:\n", df)
print("\nFiltered Data:\n", filtered)
print("\nDate Range:\n", date_range)
print("\nResampled Data (Monthly):\n", resampled)
```

```python
import pandas as pd

df = pd.read_csv('house-price.csv')
print(df.head())

print("\n\t\t\t\tinfo and summary: ")
print(df.info())
print(df.describe())
```

```python
# pd.isna(demi)
print(df.isna())
df.drop_duplicates(inplace=True)
print("\nMissing Values:\n", df.isnull().sum())

cols = ["mainroad","guestroom","basement","hotwaterheating","airconditioning","prefarea"]
for col in cols:
    df[col] = df[col].astype(str).str.strip().str.lower()
    df[col] = df[col].map({"yes": 1, "no": 2})

print(df.head())

df1 = df.sort_values(by='area',ascending=False)
df1.head()
```

```python
# Bedrooms vs Price
bedroom_price = df.groupby('bedrooms')['price'].mean()
print("\nAverage Price by Bedrooms:\n", bedroom_price)

#Furnishing status
furnish_price = df.groupby('furnishingstatus')['price'].mean()
print("\nPrice by Furnishing:\n", furnish_price)

# Parking impact
parking_price = df.groupby('parking')['price'].mean()
print("\nPrice by Parking:\n", parking_price)
# Correlation
print("\nCorrelation:\n", df.corr(numeric_only=True))


# Top 5 expensive houses
top5 = df.nlargest(5, 'price')
print("\nTop 5 Expensive Houses:\n", top5)

res = df.groupby("furnishingstatus").agg({
    "area": ["mean", "max"], 
    "price": ["mean", "max"]})
print(res)

df['price_per_sqft'] = df['price'] / df['area']
print("\nFinal Data:\n", df.head())
```
