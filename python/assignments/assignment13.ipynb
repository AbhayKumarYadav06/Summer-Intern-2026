```python
import numpy as np

arr = np.array([[6, -8, 73, -110],
                [np.nan, -8, 0, 94]])
# Replace NaN with 0
arr = np.nan_to_num(arr, nan=0)

arr_swapped = arr[:,:3].T
print(arr)
print(arr_swapped)
```

```python
import numpy as np
arr3d = np.ones((2,3,4))   
moved = np.moveaxis(arr3d, 0, -1)
print("Original shape:", arr3d.shape)
print("New shape:", moved.shape)
```

```python
import numpy as np
arr = np.array([[1, np.nan, 3],
                [4, 5, np.nan]])
col_mean = np.nanmean(arr, axis=0)
# Replace NaN with column mean
inds = np.where(np.isnan(arr))
arr[inds] = np.take(col_mean, inds[1])
print(arr)
```

```python
arr = np.array([6, -8, 73, -110])
arr = np.where(arr < 0, 0, arr)
print(arr)
```

```python
from scipy import stats

arr1 = np.array([[3, 4], [5, 6]])
arr2 = np.array([[1, 0], [7, 8]])

avg = (arr1 + arr2) / 2
mean = np.mean([arr1, arr2])
median = np.median([arr1, arr2])
mode = stats.mode([arr1, arr2], axis=None)

print("Average:\n", avg)
print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
```

```python
A = np.array([[1, -2, 3],
              [-1, 3, -1],
              [2, -5, 5]])
b = np.array([9, -6, 17])

sol = np.linalg.solve(A, b)
print("Solution (x, y, z):", sol)
```
