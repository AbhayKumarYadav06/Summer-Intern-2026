```python
import requests

def fetch_pokemon_data(pokemon_name):
    #got api key from pokeapi.co
    url = f"https://pokeapi.co/api/v2/pokemon/{pokemon_name.lower()}"
    try:
        response = requests.get(url)
        response.raise_for_status()
        data = response.json()

        print(f"POKEMON DATA: {pokemon_name.upper()} ---")
        print(f"Name: {data['name']}")
        print(f"ID: #{data['id']}")
        print(f"Height: {data['height']} decimeters")
        print(f"Weight: {data['weight']} hectograms")

        # Pokemon types
        types = [t["type"]["name"] for t in data["types"]]
        print(f"Type(s): {', '.join(types)}")
        # First 3 moves
        moves = [m["move"]["name"] for m in data["moves"][:3]]
        print(f"Some Moves: {', '.join(moves)}")

    except requests.exceptions.HTTPError:
        print(f"Error: Pokémon '{pokemon_name}' not found. Check Spelling!")
    except requests.exceptions.RequestException as e:
        print(f"Failed to fetch Pokémon data: {e}")
    print("\n")


def fetch_crypto_prices():            #got api from coingecko.com, no key needed, just a simple get request to fetch current prices of bitcoin and ethereum in USD
    url = "https://api.coingecko.com/api/v3/simple/price?ids=bitcoin,ethereum&vs_currencies=usd"

    try:
        response = requests.get(url)
        response.raise_for_status()
        data = response.json()

        print("CRYPTO MARKET PRICES:-")
        print(f"Bitcoin (BTC): ${data['bitcoin']['usd']:,} USD")
        print(f"Ethereum (ETH): ${data['ethereum']['usd']:,} USD")

    except requests.exceptions.RequestException as e:
        print(f"Failed to fetch crypto data: {e}")
    print("\n")


fetch_crypto_prices()
poke_name = input("Enter the name of a Pokémon to fetch its data: ")
fetch_pokemon_data(poke_name)
```

```python
import requests
def weather_info(city):
    url = f"https://api.openweathermap.org/data/2.5/weather?q={city}&appid=995c76dd5d56c791498d1283d6892faa&units=metric"
    try:
        response = requests.get(url)
        response.raise_for_status()  # Check if the request was successful
        data = response.json()
        print("temp: ",data['main']['temp'])
        print("pressure: ",data['main']['pressure'])
        print("Humidity: ",data['main']['humidity'])
        print("sea level: ",data['main']['sea_level'])
        print("feels like: ",data['main']['feels_like'])
    except requests.exceptions.RequestException as e:
        print(f"An error occurred: {e}")

city = input("Enter city for temprature: ")
print("Entered city: ",city)     
weather_info(city)
```

```python
#rock paper scissor game practice
import random
l = ["rock", "paper", "scissor"]
while True:
    print("Press 1 to Play AND Press 2 to Exit.")
    ch = int(input("Enter choice:"))
    
    if (ch==1):
        user_point = 0
        comp_point = 0  #variable define
        for i in range(1,7):

            user_input = input("Enter rock, paper or scissor: ")
            if user_input not in l:
                print("Invalid input. Please try again.")
                continue
            computer_input = random.choice(l)
            print(f"Computer chose: {computer_input}")
            if user_input == computer_input:
                print("It's a tie!")
            elif (user_input == "rock" and computer_input == "scissor") or (user_input == "paper" and computer_input == "rock") or (user_input == "scissor" and computer_input == "paper"):
                print("You win!")
                user_point += 1
            else:
                print("you lose.")
                comp_point += 1
        print("<GAME OVER>")
        if(user_point > comp_point):
            print("Congratulations! You won this round.")
        elif(user_point == comp_point):
            print("This round is tie! Well played.")
        else:
            print("You lose this round, Better luck next time.")

    elif(ch==2):
        print("closing the game...")
        break

    else:
        print("Invalid choice. Try again..")
```
