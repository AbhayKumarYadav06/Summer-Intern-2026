```python
import pandas as pd

#create pandas series from dict
data = {'a': 111, 'b': 222, 'c': 333, 'd': 444}
Series_dict = pd.Series(data)
print(Series_dict)
print("\n", type(Series_dict))
print("\nkey of series (a) has value: ", Series_dict['a'])

#create pandas series from lists
data = [111, 222, 333, 444]
Series_list = pd.Series(data)   
print(Series_list)
print("\n", type(Series_list))
print("\nValue of index 2 is: ", Series_list[2])

```

```python
import pandas as pd

#create pandas dataframe from 2d list
data = [['a','b',1,2], ['c','d',3,4],['e','f',5,6],['g','h',1,4]]
df = pd.DataFrame(data, columns=['col1', 'col2', 'col3', 'col4'])
print(df)

#create pandas dataframe from dict
data = {'column1': ['x', 'y', 'w', 'z'], 'column2': ['b', 'd', 'f', 'h'], 'column3': [1, 3, 5, 1], 'column4': [2, 4, 6, 4]}
df = pd.DataFrame(data)
print(df)

#create pandas dataframe from list of lists
data = [['x', 'b', 1, 2], ['y', 'd', 3, 4], ['w', 'f', 5, 6], ['z', 'h', 1, 4]]
df = pd.DataFrame(data, columns=['col1', 'col2', 'col3', 'col4'])
print(df)

#create pandas dataframe from list of tuples
data = [('x', 'b', 1, 2), ('y', 'd', 3, 4), ('w', 'f', 5, 6), ('z', 'h', 1, 4)]
df = pd.DataFrame(data, columns=['col1', 'col2', 'col3', 'col4'], index=['row1', 'row2', 'row3', 'row4'])
print(df)

#create pandas dataframe from dict of dicts
data = {'row1': {'col1': 'x', 'col2': 'b', 'col3': 1, 'col4': 2}, 
        'row2': {'col1': 'y', 'col2': 'd', 'col3': 3, 'col4': 4}, 
        'row3': {'col1': 'w', 'col2': 'f', 'col3': 5, 'col4': 6}, 
        'row4': {'col1': 'z', 'col2': 'h', 'col3': 1, 'col4': 4}}
df = pd.DataFrame(data).T  #.T is used to transpose the dataframe,,nahi to row ki jagah column or col ki jagah row aa jayega
print(df)
```

```python
import pandas as pd

data = {
    'Name': ['Abhay', 'Priya', 'Daksh', 'Shreshth', 'Tanishq'],
    'Age': [23, 21, 25, 22, 30],
    'City': ['Bhopal', 'Indore', 'Delhi', 'Bangalore', 'Jaipur']
}
df = pd.DataFrame(data)
print("Original DataFrame:\n", df)

                                #Differnt ways to iterate over rows of a dataframe

for index, row in df.iterrows():         #iterrows() har row ko ek Series ki tarah deta hai
    print(f"Index: {index} Name: {row['Name']} Age: {row['Age']}")
print("\n")

for row in df.itertuples():              #itertuples() zyada fast hota hai aur named tuples deta hai
    print(f"Index: {row.Index} Name: {row.Name} Age: {row.Age}")
print("\n")


                            #Selecting rows in pandas DataFrame based on condition
filtered_df = df[df['Age'] > 22]
print("Rows where Age > 22:\n", filtered_df)
filtered_df = df[df['City'] == 'Bhopal']
print("Rows where City is Bhopal:\n", filtered_df)
print("\n")
```

```python
#Select any row from a Dataframe using iloc[]
print(df.iloc[2])
print(df.iloc[1:4])  #rows from index 1 to 3
print("\n")


print(df.loc[1:3, ['Name', 'City']]) #Limited rows selection with given columns


#Drop rows from the dataframe based on certain condition applied on a column
df_filtered = df[df['Age'] > 22]
print(df.drop(df_filtered.index))
print(df[df['City'] != 'Mumbai'])


rows_list = df.values.tolist()            # convert dataframe se list me
print("List created from rows:\n", rows_list)


                                    # Insert new data in specific position
new_row = pd.DataFrame([{'Name': 'Kiran', 'Age': 26, 'City': 'Pune'}])
pos = 2
naya_df = pd.concat([df.iloc[:pos], new_row, df.iloc[pos:]]).reset_index(drop=True)
print("DataFrame after inserting row at index 2:\n", naya_df)
```
