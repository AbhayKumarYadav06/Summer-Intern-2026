```python
t = (1, 2, 3)
repeated_tuple = t * 3

print("Original tuple:", t)
print("Repeated tuple:", repeated_tuple)
```

```python
t1 = (10, 20)
t2 = (30, 40)
t3 = (50, 60)
combined_tuple = t1 + t2 + t3
print("Combined tuple:", combined_tuple)
```

```python
fruits = ("apple", "banana", "cherry")
search_element = "banana"

if search_element in fruits:
    print(f"found, '{search_element}' exists in the tuple.")
else:
    print(f"No, '{search_element}' does not exist in the tuple.")
```

```python
numbers = (15, 42, 7, 89, 23, 11)
total = 0
highest = numbers[0]
lowest = numbers[0]

for num in numbers:
    total += num         
    if num > highest:
        highest = num    
    if num < lowest:
        lowest = num     

print("Tuple:", numbers)
print("Total:", total)
print("Highest Value:", highest)
print("Lowest Value:", lowest)
```

```python
n = (3, 14, 7, 22, 9, 41, 18, 5)

# Filtering using a tuple comprehension/generator
filtered_n = tuple(x for x in n if x > 10)
print("Original tuple:", n)
print("Filtered tuple (values > 10):", filtered_n)
```

```python
s = {"cat", "dog", "bird", "fish"}
count = 0

for item in s:
    count += 1

print("Set:", s)
print("Number of elements:", count)
```

```python
set_a = {1, 2, 3}
set_b = {3, 4, 5}
combined_set = set_a.union(set_b)  # Alternatively: set_a | set_b

print("Combined set:", combined_set)
```

```python
s1 = {1, 2, 3, 4}
s2 = {3, 4, 5, 6}
common_elements = s1.intersection(s2)  # Alternatively: s1 & s2

print("Common elements:", common_elements)
```

```python
# Elements in either set, but not both
unique_to_each = s1.symmetric_difference(s2)  # Alternatively: s1 ^ s2

print("Elements in either but not both:", unique_to_each)
```
