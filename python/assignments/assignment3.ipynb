```python
#Write a Python function to find the maximum of three numbers.

def find_max(a, b, c):
    return max(a, b, c)
max_element = find_max(3, 7, 5)
print("The maximum element is:", max_element)
```

```python
#Write a Python function that takes a list and returns a new list with distinct elements from the first list.
def get_distinct_elements(lst):
    return list(set(lst))
original_list = [1, 2, 2, 3, 4, 4, 5]
distinct_list = get_distinct_elements(original_list)
print("Original list:", original_list)
print("Distinct elements:", distinct_list)
```

```python
# Write a Python function to multiply all the numbers in a list
def multiply_list(lst):
    result = 1
    for num in lst:
        result *= num
    print("The product of all numbers in the list is: ", result)
numbers = []
for i in range(5):
    num = int(input("Enter a number: "))
    numbers.append(num)
multiply_list(numbers)
```

```python
#Write a Python function to calculate the factorial of a number (a non-negative integer). The function accepts the number as an argument.
def factorial(n):
    if n < 0:
        return "Factorial is not defined for negative numbers."
    elif n == 0 or n == 1:
        return 1
    else:
        result = 1
        for i in range(2, n + 1):
            result *= i
        return result
number = int(input("Enter a non-negative integer: "))
fact = factorial(number)
print(f"The factorial of {number} is: ", fact)
```

```python
#Write a Python program to reverse a string.
def reverse_string(s):
    return s[::-1]
inputString = input("Enter a string: ")
result = reverse_string(inputString)
print("Input string: ", inputString)
print("Reversed string: ", result)
```

```python
# Write a Python function to check whether a number falls within a given range
def check_range(num, start, end):
    if start <= num <= end:
        return f"{num} is within the range {start} to {end}."
    else:
        return f"{num} is not within the range {start} to {end}."
number = int(input("Enter a number: "))
start_range = int(input("Enter the start of the range: "))
end_range = int(input("Enter the end of the range: "))
result = check_range(number, start_range, end_range)
print(result)
```

```python
#Write a Python function to Print Even Numbers from a Given List
def even_num_from_list(lst):
    for num in lst:
        if num % 2 == 0:
            print(num, end='   ')
numbers = []
for i in range(5):
    num = int(input("Enter a number: "))
    numbers.append(num)
print(numbers)
result = even_num_from_list(numbers)
```

```python
#Write a Python function that takes a number as a parameter and checks whether the number is prime or not.
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

number = int(input("Enter a number: "))
if is_prime(number):
    print(f"{number} is a prime number.")
else:
    print(f"{number} is not a prime number.")
```

```python
#Write a Python function that accepts a string and counts the number of upper and lower case letters
def count_upper_lower(s):
    upper_count = sum(1 for char in s if char.isupper())
    lower_count = sum(1 for char in s if char.islower())
    print(f"Number of uppercase letters: {upper_count}")
    print(f"Number of lowercase letters: {lower_count}")

input_string = input("Enter a string: ")
print("Input string: ", input_string)
count_upper_lower(input_string)
```
