```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Division by zero is not possible"
    return a / b

num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

print("Addition =", add(num1, num2))
print("Subtraction =", subtract(num1, num2))
print("Multiplication =", multiply(num1, num2))
print("Division =", divide(num1, num2))
```

```python
num = int(input("Enter a number: "))

original = num
reverse = 0

while num != 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

if original == reverse:
    print(original, "is a Palindrome Number")
else:
    print(original, "is not a Palindrome Number")
```

```python
# DICTIONARY 
students = {"Abhay": 80, "Rahul": 75, "Aman": 90}
print("Marks of Abhay:", students["Abhay"])

students["Rohit"] = 85
#Update
students["Rahul"] = 78
print("Students:", students.keys())
print("Marks:", students.values())
# Highest marks with maximum function
max_student = max(students, key=students.get)
print("Topper:", max_student, students[max_student])

#um of marks
print("Total Marks:", sum(students.values()))

#Reverse dictionary
rev = {v: k for k, v in students.items()}
print("Reversed:", rev)
#Check key
print("Is Aman present?", "Aman" in students)

#Merge dictionaries
d2 = {"Karan": 88}
students.update(d2)
print("Merged:", students)
```

```python
    # TUPLE 
t = (10, 20, 30, 40, 50)
# First & last
print("First:", t[0], "Last:", t[-1])
# Length
print("Length:", len(t))

print("Elements:", end=" ")
for i in t:
    print(i, end=" ")
print()
print("Count of 20:", t.count(20))
print("Index of 30:", t.index(30))

# Convert intto list
lst = list(t)
print("Tuple to List:", lst)
print("Max:", max(t), "Min:", min(t))   # max and min

t2 = (60, 70)
print("Joined Tuple:", t + t2)
```
