```python
stu_name = input("Name of the student: ")
stu_class = input("Class name: ")
marks_of_stu = []

for i in range(0, 5):
    marks = int(input(f"Enter the marks of subject {i+1}: "))
    marks_of_stu.append(marks)  # append() se list me marks add hote jayenge

total_marks = sum(marks_of_stu)
percentage = (total_marks / 500) * 100

# Output dekhne ke liye
print(f"\nStudent Name: {stu_name}")
print(f"Class: {stu_class}")
print(f"Marks: {marks_of_stu}")
#print(f"Total Marks: {total_marks}")
print(f"Percentage: {percentage:.2f}%")
```

```python
str1 = "Hello"
str2 = "World123"

combined = str1 + " " + str2
print(f"Combined string: '{combined}'")

print("lower():", combined.lower())              # 'hello world123'
print("upper():", combined.upper())              # 'HELLO WORLD123'
print("title():", combined.title())              # 'Hello World123'
print("swapcase():", combined.swapcase())        # 'hELLO wORLD123'
print("capitalize():", combined.capitalize())    # 'Hello world123'
print("casefold():", combined.casefold())        # 'hello world123' - more aggressive than lower()

print("center(20, '*'):", combined.center(20, '*'))  # '***Hello World123***'
print("count('l'):", combined.count('l'))            # 3
print("endswith('123'):", combined.endswith('123'))  # True
print("find('World'):", combined.find('World'))      # 6, returns -1 if not found

print("isalnum():", combined.isalnum())
print("isdigit():", combined.isdigit())
print("isnumeric():", combined.isnumeric())
print("isspace():", combined.isspace())
print("replace('or', '012'):", combined.replace('or', '012'))  # 'Hell0 World123'
```

```python
x = 10
print(f"Initial x  = {x}")

x += 5        # x = x + 5
print(f"After x += 5: {x}")

x -= 3        # x = x - 3
print(f"After x -= 3: {x}")

x *= 2        # x = x * 2
print(f"After x *= 2: {x}")

x /= 4        # x = x / 4
print(f"After x /= 4: {x}")

x = 10
x %= 3        # x = 10 % 3
print(f"After x %= 3: {x}")

x = 10
x //= 3       # x = 10 // 3
print(f"After x //= 3: {x}")

x = 2
x **= 3       # x = 2 ** 3 (2 ki power 3)
print(f"After x **= 3: {x}\n")
```

```python
subject1 = float(input("Enter marks for Subject 1: "))
subject2 = float(input("Enter marks for Subject 2: "))
subject3 = float(input("Enter marks for Subject 3: "))
subject4 = float(input("Enter marks for Subject 4: "))
subject5 = float(input("Enter marks for Subject 5: "))

total_marks = subject1 + subject2 + subject3 + subject4 + subject5
percentage = (total_marks / 500) * 100

if percentage >= 60:
    grade = 'A'
elif percentage >= 50:
    grade = 'B'
elif percentage >= 40:
    grade = 'C'
elif percentage >= 33:
    grade = 'D'
else:
    grade = 'Fail'

# Display the results
print("\n--- Results ---")
print(f"Total Marks: {total_marks} / 500")
print(f"Percentage: {percentage:.2f}%")
print(f"Grade: {grade}")
```
