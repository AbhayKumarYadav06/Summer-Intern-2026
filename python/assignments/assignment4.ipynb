```python
#Create a CSV file for address book, CSV file should have column for name, address, mobile, email. Insert 2-3 dummy data entered by user.
import csv

with open("address_book.csv", 'w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['Name', 'Address', 'Mobile', 'Email'])
    for _ in range(3):
        name = input("Enter name: ")
        address = input("Enter address: ")
        mobile = input("Enter mobile: ")
        email = input("Enter email: ")
        writer.writerow([name, address, mobile, email])
print("address_book.csv created successfully with dummy data.")
```

```python
import sqlite3
conn = sqlite3.connect('demi_db.db')
table1 = '''CREATE TABLE IF NOT EXISTS students
            (id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT NOT NULL,
            age INTEGER NOT NULL,
            grade TEXT NOT NULL);'''
conn.execute(table1)
table2 = '''CREATE TABLE IF NOT EXISTS teachers
            (id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT NOT NULL,
            subject TEXT NOT NULL);'''
conn.execute(table2)
print("database created. ")
conn.close()
```

```python
import sqlite3
conn = sqlite3.connect('demi_db.db')
sql1 = '''INSERT INTO students (name, age, grade) 
        VALUES ('Abhay', '21', 'A'),
        ('Daksh', '20', 'B'),
        ('Tanish', '22', 'A'),
        ('Shreshth', '23', 'B');'''
conn.execute(sql1)
sql2 = '''INSERT INTO teachers (name, subject) 
        VALUES ('Mr. Sharma', 'Math'),
        ('Ms. Gupta', 'Science'),
        ('Mr. Singh', 'English');'''
conn.execute(sql2)
conn.commit()
conn.close()
```

```python
import sqlite3
conn = sqlite3.connect('demi_db.db')
conn.execute("UPDATE students SET grade = 'A+' WHERE name = 'Tanish';")
conn.commit()
conn.close()
```

```python
import sqlite3
conn = sqlite3.connect('demi_db.db')
sql_1 ='''select * from students;'''
res_stu = conn.execute(sql_1)
for row in res_stu:
    print(row)
sql_2 ='''select * from teachers;'''
res_teach = conn.execute(sql_2)
for row in res_teach:
    print(row)
conn.commit()
conn.close()
```
