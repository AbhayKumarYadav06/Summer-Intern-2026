```python
#convert a series of datastrings to a time strings
import pandas as pd

s1 = pd.Series(['2026-05-27', '2026-05-28', '2026-05-29'])
res = pd.to_datetime(s1)
print(res)

s2 = pd.Series(['2026-05-27 10:30:00', '2026-05-28 12:45:00'])
res = pd.to_datetime(s2)
print(res)

s3 = pd.Series(['27-05-2026', '28-05-2026'])
res = pd.to_datetime(s3, format='%d-%m-%Y')
print(res)

res = pd.to_datetime(s2).dt.time
print(res)

s4 = pd.Series(['2026-05-27', 'wrong_data'])
res = pd.to_datetime(s4, errors='coerce')
print(res)
```

```python
import pandas as pd

df1 = pd.DataFrame({
    'id': [1,2,3,4,5],
    'name': ['A1','A2','A3','A4','A5'],
    'subject': ['sub1','sub2','sub3','sub4','sub5']    
})

df2 = pd.DataFrame({
    'id': [1,2,3,4,5],
    'name': ['B1','B2','B3','B4','B5'],
    'subject': ['sub3','sub5','sub1','sub4','sub2']    
})
#perform an inner merge on this comon column and display the resulting df
res = df1.merge(df2, on="id", how="inner")
print(res)

#perform left join of dataframe df1 and df2 on the id column. Explain the missing values resulting in df
res = df1.join(df2, on="id", lsuffix= "_left", rsuffix="_right") #nan values bcz basd on index
print(res)
res = df1.merge(df2, on="id", how="left")
print(res)

#perform right join using merge on a common column, then perform a join using df.join based on index.
res = df1.merge(df2, on="id", how="right")
print(res)

df1_new = df1.set_index("id")         #ye based on index hai
df2_new = df2.set_index("id")
res = df1_new.join(df2_new, lsuffix="_left", rsuffix="_right")
print(res)

#compare the results merging with multiple keys
res = df1.merge(df2, on=["id", "subject"], how="inner")
print(res)
```

```python
#create 3 dfs, verticaly concat two of themusing pd.concat, then merge the resulting df with third df in common key
import pandas as pd

df1 = pd.DataFrame({
    'id': [1,2],
    'name': ['A1','A2']
})

df2 = pd.DataFrame({
    'id': [3,4],
    'name': ['A3','A4']
})

df3 = pd.DataFrame({
    'id': [1,2,3,4],
    'subject': ['Math','Sci','Eng','Hindi']
})

df_concat = pd.concat([df1, df2], ignore_index=True)
print(df_concat)
res = df_concat.merge(df3, on="id", how="inner")
print(res)

```
