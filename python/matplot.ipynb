```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0,10,100)
y = np.sin(x)

plt.plot(x,y, label='sin(x)',color='red',linestyle='--')
plt.title('Sin values')
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.random.rand(50)
y = np.sin(x)

plt.scatter(x,y, label='sin(x)',color='red',alpha=0.7)
plt.title("Rrandom Values")
plt.xlabel("Y axis")
plt.ylabel("Y axis")

plt.show()
```

```python
categories = ['a','b','c','d','e','f','g','h']
values = [10,20,30,22,56,26,33,45]

plt.bar(categories,values,color='green')
plt.title("Bar Graph")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

data = np.random.rand(500)
plt.hist(data, bins=10,color='blue')
plt.title('Sin values')
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.show()
```
