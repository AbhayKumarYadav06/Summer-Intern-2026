```python
import qrcode
data = "https://www.bing.com/images/search?view=detailV2&ccid=60nlpatn&id=852C47C6D95A560139B227B3ABA6868421096181&thid=OIP.60nlpatndA3ytb7Y3bFT3gHaEK&mediaurl=https%3a%2f%2fcdn.wallpapersafari.com%2f13%2f81%2fk6Gpje.jpg&exph=1080&expw=1920&q=image+wallpaper&FORM=IRPRST&ck=DAD2BE277766E197E3D62A0FCB0F6BB4&selectedIndex=2&itb=0"
img = qrcode.make(data)
img.save('qrcode.png')
```

```python
import cv2
img = cv2.imread('qrcode.png')
detector = cv2.QRCodeDetector()
data, vertices_array, binary_qrcode = detector.detectAndDecode(img)
if vertices_array is not None and data: 
    print('QR code data: ',data)
else:
    print('QR code is not detected')
```
