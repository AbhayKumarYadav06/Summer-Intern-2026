```python
import numpy as np
a = np.array([10,20,30,40,50])
print("1d array: ",a ,type(a))
#print(a.dtype)

a = np.array([[10,20],[30,40],[50,45]], dtype='S')
print("2d array: \n",a ,type(a))

a = np.zeros([3,5])
print(a)
b = np.zeros(5)
print(b)
a = np.ones([3,6])
print(a)
b = np.ones(6)
print(b)

a = np.zeros([3,5,4])
print(a)

#fotron order arr
a= np.ones([6,3], order='F')
print(a)

arr = np.arange(10)
print("arr using arange: ",arr)
```

```python
import numpy as np
arr1 = np.linspace(0,5,num=10)
print("using linspace: ",arr1)

arr1 = np.linspace(0,5,num=10, endpoint=False)
print("using linspace: ",arr1)

arr1 = np.linspace(0,5,num=10,retstep=True)
print("using linspace: ",arr1)

#empty arr
a = np.empty([2,3]) #fill last data of memory
print("empty array: \n",a)

a = np.full((2,3),5) #fill last data of memory
print("full array: \n",a)
```

```python
import numpy as np

#   accessing the array

a= np.array([[1,2,3,4,5],[6,7,8,99,10]])
print(a[1,3])

a= np.array([
    [[1,2,3,4,5],[6,7,8,99,10]],
            [[111,12,13,14,15],[16,17,18,199,110]]
            ])
print(a[1,1,3])
print(a[1,1,-1])

#   Slicing    [start,stop,step]
s = np.array([1,22,65,78,11,12,13,14,15])
print(s[1:4:])
print(s[1:9:2])

s = np.array([[1,2,3,4,5],[6,7,8,99,10]])
print("slicing 2d arr:\n",s[0:2,1:3])

a= np.array([
    [[1,2,3,4,5],[6,7,8,99,10]],
            [[111,12,13,14,15],[16,17,18,199,110]]
            ])
print("slicing in 3d arr\n", a[0,1,2])

print(a.reshape(-1))
```

```python
print("copy wala part:")
s = np.array([1,22,65,78,11,12,13,14,15])
x = s.copy()
s[0]=56
print(s)
print(x)

print("view wala part:")
y = np.array([1,22,65,78,11,12,13,14,15])
z = s.view()
y[0]=56
print(y)
print(z)

y = np.array([1,22,65,78,11,12,13,14,15], ndmin=5)
print(y)

s = np.array([[1,2,3,6,5],[6,7,8,99,10]])
print(s.ravel())
print(np.unique(s))

print(s.max(),s.min())

print("sum: ",s.sum())
print("std deviation: ",s.std(),"and", np.std(s))
print("square root: ",np.sqrt(s))

s = s.reshape(5,2)
q = np.array([[111,12,13,14,15],[16,17,18,199,110]])
res = s.dot(q)
print(res)

```
