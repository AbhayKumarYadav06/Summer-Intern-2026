```python
l=[]
print("""
    1. Add element
    2. remove element
    3. Display first three elements
    4. Display last three elements
    5. Print
    6. Exit""")

while(True):
    ch=int(input("Enter your choice: "))
    
    if(ch==1):
        element=int(input("Enter element to add: "))
        l.append(element)
    elif(ch==2):
        if l:
            l.pop(0)
        else:
            print("Queue is empty")
    elif(ch==3):
        print(l[:3])
    elif(ch==4):
        print(l[-3:])
    elif(ch==5):    
        print(l)
    elif(ch==6):
        break   
    else:        print("Invalid choice")
```
