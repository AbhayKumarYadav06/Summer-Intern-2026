```python
import pandas as pd

data = {
    "Name": ["Abhay", "Daksh", "Shreshth", "Tanish"],
    "Subject": ["Python", "Java", "C++", "JavaScript"],
    "Marks": [85, 90, 80, 95]
}
df = pd.DataFrame(data)
print(df,"\n\n")

#remember iloc is index based and loc is label based
print(df.iloc[2],"\n")
print(df.loc[2],"\n")
print(df.loc[[0,2]],"\n")

print(df["Name"])
```

```python
import pandas as pd

data = {'day1': 'Monday', 'day2': "Tuesday", 'day3': 'Wednesday', 'day4': 'Thursday', 'day5': 'Friday'}
df1 = pd.DataFrame(list(data.items()), columns=['Day', 'Date'])
print(df1)
print(type(df1))

df = pd.Series(data, index=['day1','day3','day4'])
print(df)
print(type(df))
res = df.tolist()
print(res)
print(type(res))
```

```python
df = pd.DataFrame([['a','b',1,2], ['c','d',3,4],['e','f',5,6],['g','h',1,4]], columns=['col1', 'col2', 'col3', 'col4'])
print(df)

print("\nSlicing with df.iloc(start_row:end_row, start_col:end_col)")
print(df.iloc[1:4,0:2])
print(df.iloc[1:4,:])

print("\nAfter change index and with df.loc(start_row:end_row, start_col:end_col)")
df.index = ['row1', 'row2', 'row3', 'row4']
print(df.loc['row2':'row4','col2':'col4'])

print("\n After Modification 1.")
df.iloc[1:3,0] = ['x','y']
print(df)
print("\n After Modification 2.")
df.iloc[1:3,0:2] = [['x','y'],['z','w']]
print(df)

#removing rows and columns
res = df.drop(['row1','row3']) #axis 0 means row and axis 1 means column
#res = df.drop('row3', axis=0) #axis 0 means row and axis 1 means column
print(res)
```
