```python
import pandas as pd

#To perform arithmatic operatin df should be number type
data = {'A':[1,2,3,4,5], 'B':[10,20,30,40,50]}
df = pd.DataFrame(data)
print(df)

print("\n",df['B'][1]%3)

print("\nAddition")
print(df+5)
print("\nSubtraction: ")
print(df-10)
print("\nModulus ")
print(df%2)

df1 = df['A']*2
df1.columns = ['A']
print("\n",df1)
```

```python
#concat with pd
import pandas as pd

one = pd.DataFrame({
    'name': ['A','B','C','D'],
    'subject': ['sub1','sub2','sub3','sub4'],
    'marks': [83,85,77,92]
},index= [1,2,3,4])
print(one)

two = pd.DataFrame({
    'name': ['B','C','P','Q'],
    'subject': ['sub3','sub5','sub1','sub4'],
    'marks': [83,85,77,92]
},index= [1,2,3,4])
print(two)

res = pd.concat([one,two], keys=['x','y'])
print(res)
res1 = pd.concat([one,two]).reset_index(drop=True)
print(res1)
res2 = pd.concat([one,two],ignore_index=True)
print(res2)
res4 = pd.concat([one,two],axis=1)
print(res4)
```

```python
import pandas as pd

#merging method in pandas
#dataframe.merge(right, how="inner", on="column",left,sort=False)


one = pd.DataFrame({
    'id': [1,2,3,4,5],
    'name': ['A1','A2','A3','A4','A5'],
    'subject': ['sub1','sub2','sub3','sub4','sub5']    
})

two = pd.DataFrame({
    'id': [2,1,8,4],
    'name': ['B1','B2','B3','B4'],
    'subject': ['sub3','sub5','sub1','sub4']    
})

result = one.merge(two, on=["id","subject"])
print(result)
result = one.merge(two, on="id")
print("\n",result)


result = one.merge(two, on= ["id"], how='left')
print("\n",result)
result = one.merge(two, on= ["subject"], how='right')
print("\n",result)
result = one.merge(two, on= ["id","subject"], how='outer')
print("\n",result)
 
```

```python
import pandas as pd

#join
left = pd.DataFrame({
    'id': [1,2,3,4,5],
    'name': ['A1','A2','A3','A4','A5'],
    'subject': ['sub1','sub2','sub3','sub4','sub5']    
})

right = pd.DataFrame({
    'id': [2,1,8,4],
    'name': ['B1','B2','B3','B4'],
    'subject': ['sub3','sub5','sub1','sub4']    
})

res = left.join(right, lsuffix='left', rsuffix='right')
print(res)

res = left.join(right, lsuffix='right', rsuffix='left')
print(res)
```

```python
import pandas as pd
#pivoting -> reshaping the data

df = pd.DataFrame({
    "col1": range(12), 
    "col2": ['a']*3 + ['b']*3 + ['c']*3 + ['d']*3,
    "date": pd.to_datetime(['2026-5-22', '2026-5-21', '2026-5-20']*4)
})
print(df)

print("time to pivot.\n")
pivoted = pd.pivot(df, index='date', columns="col2", values="col1")
print(pivoted)
```

```python
#pivot table
import pandas as pd
df = pd.DataFrame({
    'course': ['cs','ds','ai','ece'],
    'student': ['A','B','C','D'],
    'year': [1,2,3,1]
})
df['marks'] = df['year'][:3]*45
print(df)
pivoted = pd.pivot_table(df, index='course', columns='year', values='student', aggfunc='sum')
print(pivoted)
```
