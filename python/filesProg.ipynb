```python
import csv

d = [
    ["Abhay", 21, "New delhi"],
    ["tanish", 25, "jaipur"],
    ["priya", 28, "mumbai"],
    ["daksh", 325, "bangalore"]
]

with open("fry.csv", "w", newline="") as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(["Name", "Age", "City"])
    for row in d:
        writer.writerow(row)


with open("fry.csv", "a", newline="") as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(["Shreshth", 27, "kolkata"])
    writer.writerow(["Ananya", 22, "chennai"])

with open("fry.csv", "r") as csvfile:
    reader = csv.reader(csvfile)
    for i in reader:
        print(i)
```
