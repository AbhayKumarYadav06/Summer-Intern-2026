```python
file = open("try.txt", "w")
file.write("This is a file created using Python.")
file.write(" It demonstrates file handling in Python.")
file.close()
```

```python
file = open("try.txt", "r")
content = file.read()
print(content)
file.close()
```

```python
file = open("try.txt", "a")
file.write(" This line is appended to the file.\n here we are adding a new line to the existing content.")
file.close()
```

```python
file = open("try.txt", "r")
for x in file:
    print(x)
file.close()
```
