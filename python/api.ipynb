```python
import requests
def weather_info(city):
    url = f"https://api.openweathermap.org/data/2.5/weather?q={city}&appid=995c76dd5d56c791498d1283d6892faa&units=metric"
    try:
        response = requests.get(url)
        response.raise_for_status()  # Check if the request was successful
        data = response.json()
        print("temp: ",data['main']['temp'])
        print("pressure: ",data['main']['pressure'])
        print("Humidity: ",data['main']['humidity'])
        print("sea level: ",data['main']['sea_level'])
        print("feels like: ",data['main']['feels_like'])
    except requests.exceptions.RequestException as e:
        print(f"An error occurred: {e}")

city = input("Enter city for temprature: ")
print("Entered city: ",city)     
weather_info(city)
```
