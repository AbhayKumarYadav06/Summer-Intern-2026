```python
#recursion
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)
print(factorial(5))

#iterative
def factorial_iterative(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result
print(factorial_iterative(5))
```

```python
def rec(k):
    if k > 0:
        temp = k + rec(k - 1)
        print(temp)
    else:
        temp = 0
    return temp
print("\nRecursion Example:->")
rec(6)
```
