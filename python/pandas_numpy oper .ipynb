```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "col1": [2, np.nan, np.nan, 4], 
    "col2": [1, pd.NA, pd.NA, 2]
})
print(df)

#Fill missing values with hyphen(-)
f = df.fillna("-")
print(f)

print("ye previous values fil karega missing me\n")
res = df.ffill()
print(res)

print("ye next values fil karega missing me\n")
res = df.bfill()
print(res)
```

```python
import pandas as pd
df = pd.DataFrame(
    [[12,3,11],[-1,-2,8],[7,8,9]],
    index=['a','c','d'],
    columns=['one', 'two', 'three']
)
print(df)
df = df.reindex(['a','b','z','c','d','e','f','g'])
print(df)

print("\nlinit in filling forward se aage 'n' times fill karega ")
res = df.ffill(limit=1)
print(res)
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "one": ['a', 'b', 'a', 'c', 'a' ,'c', 'b'], 
    "two": [1, 88, 20, 20, 78, 88, 56]
})
print(df,"\n\n")

grouped_data = df.groupby('one')['two'].sum()
print(grouped_data)
grouped_data = df.groupby('one')['two'].mean()
print(grouped_data)
grouped_data = df.groupby('one')['two'].count()
print(grouped_data)
```

```python
import pandas as pd
df = pd.DataFrame({
    'name': ['abhay','tanish','daksh','shreshth'],
    'age': [20,21,22,20]
}, index=[1,3,2,5])
print(df)
res =df.sort_values(by='age',ascending=False)  #descending order
print(res)
res =df.sort_values(by='age',ascending=True)   #ascending order
print(res)

print("\ninplace:")
df.sort_values(by='age', inplace=True) 
print(df) 

print("\nsort index:")
res = df.sort_index() 
print(res)
```
