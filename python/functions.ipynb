```python
def functionname():
    print("Hello, sir kaise ho?")

functionname()
```

```python
def func_sum(a, b):
    return a + b

result = func_sum(10, 20)
print(result)
```

```python
#arbitary positional arguments
def summation(*args):
    total = 0
    for num in args:
        total += num
    return total
result = summation(1, 2, 3, 4, 5)
print(result)



def cities(*city):
    print("The cities are:", city)
cities("Delhi", "Mumbai", "Bangalore")
```

```python
#arbitary keyword arguments
def city(**citys):
    print("The cities are:", citys)
city(Delhi="India", Mumbai="India", Bangalore="India")
```

```python
#default parameter values
def greet(name="Guest"):
    print("Welcome,", name)

greet("Abhay")
greet()
```

```python
#passing a list as an argument
def printList(items):
        print("list has items: \n",items)

my_list = ["apple", "banana", "cherry"]
printList(my_list)
```

```python
#passing a function as an argument
def doubleTheValue(x):
    return x * 2

def apply_operation(func, value):
    return func(value)

result = apply_operation(doubleTheValue, 555)
print(result)
```

```python
#only positional arguments
def doubleValue(x, /):
    return x * 2
result = doubleValue(10)
#doubleValue(x=10)  # This will raise an error because x is a positional-only argument
print(result)

#only keyword arguments
def greet(*, name):
    print("Hello,", name)
greet(name="Abhay")
#greet("Alice")  # This will raise an error because name is a keyword-only argument

#combined positional and keyword arguments
def my_function(a, b,/, *, c, d):
    print("sum: ", a+b+c+d)
my_function(1, 2, c=3, d=4)
```
