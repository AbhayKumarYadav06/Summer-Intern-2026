```python
s = {"set", 10, "python", "programming", 1, "set"}  #no duplicates allowed
s1 = {"abc", 10, 20}
s.add("suraj")
print(s, type(s))

s2 = s.union(s1)  #union (s|s1)
s3 = s.intersection(s1)  #intersection (s&s1)
s4 = s.difference(s1)  #difference (s-s1)

print(s2)
print(s3)   
print(s4)
s.remove(10)  #removes 10
s1.clear()  #removes all elements
print(s)
print(s1)
```
